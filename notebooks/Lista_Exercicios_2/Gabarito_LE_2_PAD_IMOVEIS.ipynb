## Parte 1 — Gabarito das Questões de Código (Q1–Q20)


### ✅ Questão 1 — 0,40 pt | **Inspeção Inicial — Shape e Describe**


```python
# (a) Dimensões do DataFrame (linhas, colunas)
print(df.shape)

# (b) Estatísticas descritivas das colunas numéricas
df.describe()
```

### ✅ Questão 2 — 0,40 pt | **Filtro Simples por Igualdade**


```python
resultado = df[df['situacao'] == 'Disponível']
resultado[['id_imovel', 'bairro', 'tipo', 'valor_venda']]
```

### ✅ Questão 3 — 0,40 pt | **Criação de Nova Coluna Derivada**


```python
# preco_m2 = valor_venda / area_m2
df['preco_m2'] = df['valor_venda'] / df['area_m2']
df[['tipo', 'area_m2', 'valor_venda', 'preco_m2']].head()
```

### ✅ Questão 4 — 0,40 pt | **Máscara Booleana Composta com `&`**


```python
mascara = (df['tipo'] == 'Apartamento') & (df['quartos'] >= 3)
resultado = df[mascara][['id_imovel', 'bairro', 'quartos', 'valor_venda']]
print(f"Imóveis encontrados: {len(resultado)}")
resultado
```

### ✅ Questão 5 — 0,40 pt | **Tratamento de Valores Ausentes — Mediana e Constante**


```python
df_trat = df.copy()

# valor_condominio → mediana global
df_trat['valor_condominio'] = df_trat['valor_condominio'].fillna(df_trat['valor_condominio'].median())

# vagas_garagem → constante 1
df_trat['vagas_garagem'] = df_trat['vagas_garagem'].fillna(1)

print(df_trat[['valor_condominio', 'vagas_garagem']].isnull().sum())
```

### ✅ Questão 6 — 0,40 pt | **Conversão de Datas e Extração de Componentes**


```python
df['data_anuncio'] = pd.to_datetime(df['data_anuncio'], format='%d/%m/%Y')
df['mes_anuncio']   = df['data_anuncio'].dt.month
df['trimestre']     = df['data_anuncio'].dt.quarter
df[['id_imovel', 'data_anuncio', 'mes_anuncio', 'trimestre']].head()
```

### ✅ Questão 7 — 0,40 pt | **Filtro com `.isnull()` — Valores Ausentes**


```python
# Identifique imóveis com valor_iptu_anual ausente
sem_iptu = df[df['valor_iptu_anual'].isnull()]
print(f"Imóveis sem IPTU cadastrado: {len(sem_iptu)}")
sem_iptu[['id_imovel', 'bairro', 'tipo', 'situacao']]
```

### ✅ Questão 8 — 0,40 pt | **Criação de Coluna com `np.where`**


```python
# Classifique como 'Alto Padrão' se valor_venda > 1_500_000, senão 'Padrão'
df['categoria_preco'] = np.where(
    df['valor_venda'] > 1_500_000,   # condição
    'Alto Padrão',                    # valor se True
    'Padrão'                          # valor se False
)
df[['id_imovel', 'tipo', 'valor_venda', 'categoria_preco']].head(8)
```

### ✅ Questão 9 — 0,40 pt | **Agrupamento com `.agg()` — Nomes Personalizados**


```python
resumo_tipo = df.groupby('tipo').agg(
    media_valor   = ('valor_venda', 'mean'),
    total_imoveis = ('id_imovel',   'count'),
    media_area    = ('area_m2',     'mean')
).round(0)
resumo_tipo
```

### ✅ Questão 10 — 0,40 pt | **Pivot Table — Valor Médio por Bairro e Tipo**


```python
pivot = pd.pivot_table(
    df,
    values='valor_venda',
    index='bairro',
    columns='tipo',
    aggfunc='mean',
    fill_value=0
).round(0)
pivot
```

### ✅ Questão 11 — 0,40 pt | **Seleção com `.loc` e Renomeação**


```python
selecionado = df.loc[df['situacao'] == 'Disponível', ['id_imovel', 'bairro', 'tipo', 'valor_venda', 'area_m2']]
selecionado = selecionado.rename(columns={'valor_venda': 'preco_anunciado'})
selecionado.head(8)
```

### ✅ Questão 12 — 0,40 pt | **Filtro com `.isin()`**


```python
bairros_nobres = ['Jardins', 'Itaim Bibi', 'Moema']
resultado = df[df['bairro'].isin(bairros_nobres)]
print(f"Imóveis nos bairros selecionados: {len(resultado)}")
resultado[['id_imovel', 'bairro', 'tipo', 'valor_venda']].head(10)
```

### ✅ Questão 13 — 0,40 pt | **Ordenação com Múltiplos Critérios**


```python
resultado = df.sort_values(
    ['bairro', 'valor_venda'],
    ascending=[True, False]
)[['id_imovel', 'bairro', 'tipo', 'valor_venda', 'area_m2']]
resultado.head(10)
```

### ✅ Questão 14 — 0,40 pt | **Operações com Strings — `.str`**


```python
# (a) bairro em maiúsculas
df['bairro_upper'] = df['bairro'].str.upper()

# (b) verifica se bairro contém 'Vila'
df['eh_vila'] = df['bairro'].str.contains('Vila')

# (c) comprimento do nome do bairro
df['len_bairro'] = df['bairro'].str.len()

df[['bairro', 'bairro_upper', 'eh_vila', 'len_bairro']].head(8)
```

### ✅ Questão 15 — 0,40 pt | **Aplicar Função com `.apply()`**


```python
def classifica_porte(area):
    if area < 60:
        return 'Compacto'
    elif area < 130:
        return 'Médio'
    else:
        return 'Grande'

df['porte'] = df['area_m2'].apply(classifica_porte)
df[['tipo', 'area_m2', 'porte']].head(8)
```

### ✅ Questão 16 — 0,40 pt | **Contagem e Distribuição Percentual**


```python
# (a) Contagem por situacao
contagem = df['situacao'].value_counts()
print("Contagem por situação:")
print(contagem)

# (b) Distribuição percentual por tipo
pct = df['tipo'].value_counts(normalize=True) * 100
print("\nDistribuição por tipo (%):")
print(pct.round(1))
```

### ✅ Questão 17 — 0,40 pt | **Top 3 Bairros por Valor Médio**


```python
top3_bairros = df.groupby('bairro').agg(
    valor_medio  = ('valor_venda', 'mean'),
    qtd_imoveis  = ('id_imovel',  'count'),
    area_media   = ('area_m2',    'mean')
).sort_values('valor_medio', ascending=False).head(3)
top3_bairros.round(0)
```

### ✅ Questão 18 — 0,40 pt | **Variável Ordinal + Discretização com `pd.cut`**


```python
# (a) Categorical ordinal: Regular < Bom < Ótimo
ordem = ['Regular', 'Bom', 'Ótimo']
df['avaliacao_corretor'] = pd.Categorical(
    df['avaliacao_corretor'], categories=ordem, ordered=True
)

# (b) pd.cut em valor_venda
bins   = [0, 700_000, 1_500_000, 3_000_000, 6_000_000]
labels = ['Econômico', 'Médio', 'Alto Padrão', 'Luxo']
df['faixa_valor'] = pd.cut(
    df['valor_venda'], bins=bins, labels=labels, include_lowest=True
)

# (c) Contagem por faixa
df.groupby('faixa_valor', observed=True)['id_imovel'].count()
```

### ✅ Questão 19 — 0,40 pt | **Empilhamento de DataFrames com `pd.concat`**


```python
df_novos = pd.DataFrame({
    'id_imovel': [131, 132],
    'data_anuncio': ['15/12/2024', '20/12/2024'],
    'bairro': ['Moema', 'Jardins'],
    'tipo': ['Apartamento', 'Cobertura'],
    'quartos': [2, 4],
    'area_m2': [80.0, 300.0],
    'valor_venda': [790000., 3800000.],
    'valor_condominio': [1100., 6000.],
    'valor_iptu_anual': [3900., 22000.],
    'andar': [9., 28.],
    'vagas_garagem': [1., 4.],
    'situacao': ['Disponível', 'Disponível'],
    'avaliacao_corretor': ['Ótimo', 'Ótimo'],
    'aceita_financiamento': [True, False],
})

df_total = pd.concat([df, df_novos], axis=0, ignore_index=True)
print(f"Shape original : {df.shape}")
print(f"Shape após concat: {df_total.shape}")
df_total.tail(3)
```

### ✅ Questão 20 — 0,40 pt | **Combinação de DataFrames com `pd.merge`**


```python
df_corretores = pd.DataFrame({
    'bairro':       ['Moema', 'Pinheiros', 'Itaim Bibi', 'Vila Madalena', 'Jardins', 'Brooklin'],
    'corretor':     ['Ana Silva', 'João Pereira', 'Maria Santos',
                     'Carlos Oliveira', 'Fernanda Lima', 'Roberto Costa'],
    'comissao_pct': [0.05, 0.045, 0.05, 0.04, 0.055, 0.04],
})

df_merged = pd.merge(
    df,
    df_corretores,
    on='bairro',
    how='left'
)
df_merged[['id_imovel', 'bairro', 'tipo', 'valor_venda', 'corretor', 'comissao_pct']].head(8)
```
