### Questão 12 🟡 MÉDIO — 0.6 pt

**Tratamento de Valores Ausentes — fillna + média**

Complete o código para tratar os valores ausentes: preencha `desconto_pct` com `0.0` e `quantidade` com a **média** da coluna.

```python
df_trat = df.copy()

# Preencha desconto_pct com 0.0
df_trat['desconto_pct'] = df_trat['desconto_pct'].fillna(0.0)

# Preencha quantidade com a média da coluna
df_trat['quantidade'] = df_trat['quantidade'].fillna(df_trat['quantidade'].mean())

print(df_trat.isnull().sum())
```

### Questão 16 🔴 DIFÍCIL — 0.8 pt

**Pivot Table**

Complete o código para criar uma pivot table com a **média** de `valor_total` por `regiao` (linhas) e `categoria` (colunas).

### Questão 17 🔴 DIFÍCIL — 0.8 pt

**Agregação com .agg() e Nomes Personalizados**

Complete o script usando `.agg()` para calcular, por `categoria`: a **soma** de `valor_total` (→ `total_vendas`), a **média** de `preco_unitario` (→ `preco_medio`) e a **contagem** de pedidos (→ `num_pedidos`).

```python
resumo = df.groupby('categoria').agg(
    total_vendas = ('valor_total',    'sum'),
    preco_medio  = ('preco_unitario', 'mean'),
    num_pedidos  = ('id_pedido',      'count')
)
resumo
```

### Questão 18 🔴 DIFÍCIL — 0.8 pt

**Variável Categórica Ordinal e Filtro Comparativo**

Complete o código para: (a) converter `nivel_satisfacao` em variável categórica **ordinal** com a ordem `Ruim < Regular < Bom < Excelente`; (b) filtrar apenas pedidos com satisfação **maior que** `'Regular'`.

```python
ordem = ['Ruim', 'Regular', 'Bom', 'Excelente']

df['nivel_satisfacao'] = pd.Categorical(
    df['nivel_satisfacao'],
    categories=ordem,
    ordered=True
)

# Filtrar apenas satisfação > 'Regular'
df_filtrado = df[df['nivel_satisfacao'] > 'Regular']
df_filtrado[['produto', 'nivel_satisfacao']]
```

### Questão 19 🔴 DIFÍCIL — 0.8 pt

**Discretização por Frequência — pd.qcut()**

Complete o código para discretizar `preco_unitario` em **4 faixas de igual frequência** usando `pd.qcut()`, com labels `['Econômico', 'Padrão', 'Premium', 'VIP']`. Em seguida, exiba a contagem de pedidos em cada faixa.

```python
df['faixa_preco'] = pd.qcut(
    df['preco_unitario'],
    q=4,
    labels=['Econômico', 'Padrão', 'Premium', 'VIP'],
    duplicates='drop'
)
print(df['faixa_preco'].value_counts())
```

### Questão 20 🔴 DIFÍCIL — 0.8 pt

**Percentual de Ausentes + Discretização Manual — pd.cut()**

Complete o script que: **(a)** calcula o percentual de ausentes por coluna; **(b)** exibe apenas colunas com ausência > 0%; **(c)** cria faixas manuais de `preco_unitario` com `pd.cut()` usando fronteiras `[0, 1000, 2000, 3000, 5000]` e labels `['Baixo', 'Médio', 'Alto', 'Premium']`.

```python
# (a) Percentual de ausentes
pct_ausentes = (df.isnull().sum() / len(df) * 100).round(2)

# (b) Exibir apenas colunas com ausência > 0%
print(pct_ausentes[pct_ausentes > 0])

# (c) Discretizar preco_unitario com fronteiras manuais
df['faixa_manual'] = pd.cut(
    df['preco_unitario'],
    bins=[0, 1000, 2000, 3000, 5000],
    labels=['Baixo', 'Médio', 'Alto', 'Premium'],
    include_lowest=True
)
df[['preco_unitario', 'faixa_manual']].head(8)
```
