##  Operações com Colunas e Linhas


- **Criando Novas Colunas**: Uma das operações mais comuns é criar uma nova coluna a partir de outras existentes.
    - **Exemplo**: Calcular o valor total de cada venda (Quantidade * Preço_Unitario).

````python
# Criando uma nova coluna 'Valor_Total'
df_vendas['Valor_Total'] = df_vendas['Quantidade'] * df_vendas['Preco_Unitario']
print("DataFrame com a nova coluna 'Valor_Total':")
display(df_vendas)
````

## Merge, Join e Concatenação

- Introdução: Combinar DataFrames é fundamental para a análise de dados.
- O pandas oferece três funções principais para isso: concat, merge e join.
    - **pd.concat()**: "Empilha" DataFrames, seja verticalmente (novas linhas) ou horizontalmente (novas colunas).
    - **pd.merge()**: Combina DataFrames com base em valores de colunas comuns (chaves), similar aos JOINs de SQL.
        - Tipos de merge: inner (padrão), outer, left, right.
    - .join(): Uma forma mais simples de combinar DataFrames com base nos seus índices.

````python
# Combinando df_vendas e df_vendedores usando uma chave comum 'ID_Vendedor'
# O 'how="left"' garante que todas as vendas sejam mantidas, mesmo que um vendedor não esteja na outra tabela.
df_completo = pd.merge(df_vendas, df_vendedores, on='ID_Vendedor', how='left')
print("DataFrame combinado (vendas com informações dos vendedores):")
display(df_completo)
````

##  Agrupamento e Agregação

- O que é Agrupamento? É o processo de dividir os dados em grupos com base em algum critério. No pandas, usamos o método .groupby()
- O Processo "Split-Apply-Combine":
    - Split: Os dados são divididos em grupos.
    - Apply: Uma função é aplicada a cada grupo independentemente (soma, média, contagem, etc.).
    - Combine: Os resultados são combinados em uma nova estrutura de dados.
- Funções de Agregação Comuns: .sum(), .mean(), .count(), .max(), .min().
- Agregações Múltiplas: O método .agg() permite aplicar várias funções de uma só vez.

- Agrupar por Nome_Vendedor e somar o Valor_Total vendido por cada um:

````python
df_completo.groupby("Nome_Vendedor")["Valor_Total"].sum()
````

- Agrupar por duas chaves (Produto, Regiao) e contar linhas:

````python
df_completo.groupby(["Produto", "Regiao"]).size()
````

### Processo Split–Apply–Combine

- **Objetivo**: obter, para cada Produto, a soma da Quantidade e a média do Preco_Unitario.

- **Split**: dividir por Produto

- **Apply**: aplicar soma e média

- **Combine**: juntar os resultados em um DataFrame

````python
(
    df_completo
      .groupby("Produto")
      .agg(Quantidade_total=("Quantidade", "sum"),
           Preco_medio=("Preco_Unitario", "mean"))
      .reset_index()
)
````

### Funções de agregação comuns
a) `.sum()` — quantidade total por Produto

````python
df_completo.groupby("Produto")["Quantidade"].sum()
````

b) `.mean()` — preço unitário médio por Produto

````python
df_completo.groupby("Produto")["Preco_Unitario"].mean()
````

c) `.count()` — número de vendas por Regiao

````python
df_completo.groupby("Regiao")["ID_Vendedor"].count()
````

d) `.max()` e `.min()` — maior e menor Valor_Total por Nome_Vendedor

````python
df_completo.groupby("Nome_Vendedor")["Valor_Total"].agg(["max", "min"])
````

### Agregações múltiplas com `.agg()`

a) Usando dicionário (gera colunas multi-índice)

````python
df_completo.groupby("Produto").agg({
    "Quantidade": ["sum", "mean"],
    "Valor_Total": ["sum", "max"],
})
````

b) Usando Named Aggregations (colunas já “achatadas” e nomeadas)

````python
df_completo.groupby("Produto").agg(
    Qtde_total=("Quantidade", "sum"),
    Qtde_media=("Quantidade", "mean"),
    Receita_total=("Valor_Total", "sum"),
    Receita_max=("Valor_Total", "max"),
).reset_index()
````

c) Transform (mesma forma do original; útil para criar colunas “por grupo”):

````python
df_completo["pct_da_receita_do_vendedor"] = (
    df_completo["Valor_Total"] /
    df_completo.groupby("Nome_Vendedor")["Valor_Total"].transform("sum")
)
df_completo.head()
````

## Pivot Tables e Reshaping

- **Remodelar (Reshaping)**: Alterar a estrutura de um DataFrame para uma melhor visualização ou análise.
- **Tabelas Dinâmicas (pivot_table)**: Uma de resumir e reorganizar dados, similar às tabelas dinâmicas do Excel.

#### Componentes Principais:
- `index`: O que se tornará o novo índice do DataFrame.
- `columns`: Os valores desta coluna se tornarão as novas colunas.
- `values`: Os valores a serem agregados.
- `aggfunc`: A função de agregação a ser aplicada (padrão é a média).

````python
# Criando uma tabela dinâmica para ver o total de vendas por vendedor em cada região
pivot_vendas_regiao = df_completo.pivot_table(index='Nome_Vendedor',
                                              columns='Regiao',
                                              values='Valor_Total',
                                              aggfunc='sum',
                                              fill_value=0) # Preenche NaN com 0

print("Tabela dinâmica de Vendas por Vendedor e Região:")
display(pivot_vendas_regiao)
````
