### Questão 3 — 0,75 pt

**Criação de Coluna Derivada**

Crie a coluna `valor_bruto` = `quantidade` × `preco_unitario`.

> ⚠️ **Atenção:** esta coluna será usada nas questões Q7 e Q9 — execute esta célula antes de continuar.

### Questão 4 — 0,75 pt

**Máscara Booleana Composta com `|` (OU)**

Complete o código para filtrar pedidos da categoria `'Eletrônicos'` **OU** com `preco_unitario >= 800`.

```python
mascara = (df['categoria'] == 'Eletrônicos') | (df['preco_unitario'] >= 800)
resultado = df[mascara][['id_pedido', 'categoria', 'produto', 'preco_unitario']]
print(f"Pedidos encontrados: {len(resultado)}")
resultado
```

### Questão 5 — 0,75 pt

**Tratamento de Valores Ausentes**

Complete o código para tratar os NaN: `desconto_pct` com `0.0` e `quantidade` com a **mediana** global da coluna.

```python
df_trat = df.copy()

# desconto_pct → 0.0
df_trat['desconto_pct'] = df_trat['desconto_pct'].fillna(0.0)

# quantidade → mediana global
df_trat['quantidade'] = df_trat['quantidade'].fillna(df_trat['quantidade'].median())

print(df_trat[['desconto_pct', 'quantidade']].isnull().sum())
```

### Questão 6 — 0,75 pt

**Conversão de Datas e Extração de Componentes**

Complete o código para: (1) converter `data_pedido` para `datetime`; (2) extrair o **ano** em `ano_pedido`; (3) extrair o **dia da semana** em `dia_semana` (0 = Segunda, 6 = Domingo).

```python
df['data_pedido'] = pd.to_datetime(df['data_pedido'], format='%d/%m/%Y')
df['ano_pedido']  = df['data_pedido'].dt.year
df['dia_semana']  = df['data_pedido'].dt.dayofweek
df[['produto', 'data_pedido', 'ano_pedido', 'dia_semana']].head()
```

### Questão 7 — 0,75 pt

**Agrupamento com `.agg()` e Ordenação**

Complete o código para agrupar por `categoria` e calcular, em uma única chamada `.agg()`, a **soma** de `valor_bruto` (→ `total_vendas`) e a **contagem** de `id_pedido` (→ `qtd_pedidos`). Ordene pelo total de vendas de forma **decrescente**.

```python
resumo = df.groupby('categoria').agg(
    total_vendas = ('valor_bruto', 'sum'),
    qtd_pedidos  = ('id_pedido',   'count')
).sort_values('total_vendas', ascending=False)
resumo
```

### Questão 8 — 0,75 pt

**Seleção com `.loc` e Renomeação**

Use `.loc` para selecionar as colunas `produto`, `categoria`, `preco_unitario` e `avaliacao` dos pedidos com `avaliacao == 'Excelente'`. Em seguida, renomeie `preco_unitario` para `'preço'`.

```python
selecionado = df.loc[
    df['avaliacao'] == 'Excelente',
    ['produto', 'categoria', 'preco_unitario', 'avaliacao']
]
selecionado = selecionado.rename(columns={'preco_unitario': 'preço'})
selecionado
```

### Questão 9 — 0,75 pt

**Pivot Table — Preço Médio por Categoria e Status**

Complete o código para criar uma pivot table com o **preço médio** (`preco_unitario`) por `categoria` (linhas) e `status` (colunas). Use `fill_value=0` e arredonde para 2 casas decimais.

```python
pivot = pd.pivot_table(
    df,
    values='preco_unitario',
    index='categoria',
    columns='status',
    aggfunc='mean',
    fill_value=0
).round(2)
pivot
```

### Questão 10 — 0,75 pt

**Variável Ordinal + Discretização Manual + Análise por Faixa**

Complete o script em três partes:

**(a)** Converta `avaliacao` para categórica **ordinal** (ordem: `Ruim < Regular < Bom < Excelente`)

**(b)** Discretize `preco_unitario` em **4 faixas manuais** com `pd.cut()` usando os limites `[0, 200, 500, 1200, 4000]` e labels `['Econômica', 'Intermediária', 'Premium', 'Luxo']`

**(c)** Calcule a **média de `preco_unitario`** por `faixa_preco` com `groupby`

```python
# (a) Categorical ordinal
ordem = ['Ruim', 'Regular', 'Bom', 'Excelente']
df['avaliacao'] = pd.Categorical(
    df['avaliacao'], categories=ordem, ordered=True
)

# (b) Faixas manuais com pd.cut
df['faixa_preco'] = pd.cut(
    df['preco_unitario'],
    bins=[0, 200, 500, 1200, 4000],
    labels=['Econômica', 'Intermediária', 'Premium', 'Luxo'],
    include_lowest=True
)

# (c) Média de preco_unitario por faixa
df.groupby('faixa_preco', observed=True)['preco_unitario'].mean().round(2)
```
