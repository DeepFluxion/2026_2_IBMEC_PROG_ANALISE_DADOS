### Questão 3 — 0,75 pt

### Questão 4 — 0,75 pt

**Máscara Booleana Composta com `&`**

Complete o código para filtrar funcionários **simultaneamente** do departamento `'TI'` **E** com `salario >= 8000`.

> 💡 **Dica de sintaxe:** Cada condição entre parênteses: `(cond1) & (cond2)`; nunca use `and`/`or` em Series pandas.

```python
mascara = (df['departamento'] == 'TI') & (df['salario'] >= 8000)
resultado = df[mascara][['id_func', 'cargo', 'nivel_cargo', 'salario']]
print(f"Funcionários encontrados: {len(resultado)}")
resultado
```

### Questão 5 — 0,75 pt

**Tratamento de Valores Ausentes — Mediana e Média**

Preencha os NaN: `bonus_pct` com a **mediana global** e `horas_extras` com a **média global**.

> 💡 **Dica de sintaxe:** `serie.fillna(serie.median())`; `serie.fillna(serie.mean())`; trabalhe sobre `df.copy()`.

```python
df_trat = df.copy()

# bonus_pct → mediana global
df_trat['bonus_pct']    = df_trat['bonus_pct'].fillna(df_trat['bonus_pct'].median())

# horas_extras → média global
df_trat['horas_extras'] = df_trat['horas_extras'].fillna(df_trat['horas_extras'].mean())

print(df_trat[['bonus_pct', 'horas_extras']].isnull().sum())
```

### Questão 6 — 0,75 pt

**Conversão de Datas e Feature Engineering**

Converta `data_admissao` para datetime; extraia `mes_admissao`; calcule `anos_empresa` (dias // 365).

> 💡 **Dica de sintaxe:** `pd.to_datetime(col, format='%d/%m/%Y')`; `.dt.month` / `.dt.year`; diferença → `.dt.days`.

```python
df['data_admissao'] = pd.to_datetime(df['data_admissao'], format='%d/%m/%Y')
df['mes_admissao']  = df['data_admissao'].dt.month
df['anos_empresa']  = (pd.Timestamp.today() - df['data_admissao']).dt.days // 365
df[['cargo', 'data_admissao', 'mes_admissao', 'anos_empresa']].head()
```

### Questão 7 — 0,75 pt

**Agrupamento + Ordenação e `.loc` + `rename`**

**(a)** `groupby` para salário médio por departamento, decrescente.

**(b)** `.loc` para `avaliacao == 'Excelente'`, colunas cargo/nivel_cargo/salario; renomeie `salario` → `remuneração`.

> 💡 **Dica de sintaxe:** `groupby('col')['col2'].mean().sort_values(ascending=False)`; `.loc[mascara, cols]`; `.rename(columns={'antigo':'novo'})`.

```python
# (a) Salário médio por departamento (maior → menor)
media_dep = df.groupby('departamento')['salario'].mean().sort_values(ascending=False)
print(media_dep)

# (b) .loc + rename
selecionado = df.loc[df['avaliacao'] == 'Excelente', ['cargo', 'nivel_cargo', 'salario']]
selecionado = selecionado.rename(columns={'salario': 'remuneração'})
selecionado
```

### Questão 8 — 0,75 pt

**Pivot Table — Salário Médio**

Pivot table com salário médio por `departamento` (linhas) × `nivel_cargo` (colunas). `fill_value=0`, arredondado para 0 casas.

> 💡 **Dica de sintaxe:** `pd.pivot_table(df, values=, index=, columns=, aggfunc=, fill_value=)`.

```python
pivot = pd.pivot_table(
    df,
    values='salario',
    index='departamento',
    columns='nivel_cargo',
    aggfunc='mean',
    fill_value=0
).round(0)
pivot
```

### Questão 9 — 0,75 pt

**Agregação com `.agg()` — Nomes Personalizados**

Por `departamento`: soma de `salario` (→ `total_salarios`), média de `horas_extras` (→ `media_horas_ext`), contagem de `id_func` (→ `num_funcionarios`).

> 💡 **Dica de sintaxe:** `.agg(nome=('coluna_origem', 'função'))` — `'sum'`, `'mean'`, `'count'`…

```python
resumo = df.groupby('departamento').agg(
    total_salarios   = ('salario',      'sum'),
    media_horas_ext  = ('horas_extras', 'mean'),
    num_funcionarios = ('id_func',      'count')
).round(2)
resumo
```

### Questão 10 — 0,75 pt

**Variável Ordinal + Discretização + Análise por Faixa**

**(a)** `pd.Categorical` ordinal para `nivel_cargo` (Júnior < Pleno < Sênior < Especialista). **(b)** `pd.qcut` em `salario`, 4 faixas, labels `['Baixo','Médio','Alto','Topo']`. **(c)** Média salarial por `faixa_salario` com `groupby`.

> 💡 **Dica de sintaxe:** `pd.Categorical(s, categories=[...], ordered=True)`; `pd.qcut(col, q=N, labels=[...])`; `groupby('col', observed=True)`.

```python
# (a) Categorical ordinal
ordem = ['Júnior', 'Pleno', 'Sênior', 'Especialista']
df['nivel_cargo'] = pd.Categorical(
    df['nivel_cargo'], categories=ordem, ordered=True
)

# (b) Discretização por frequência
df['faixa_salario'] = pd.qcut(
    df['salario'], q=4,
    labels=['Baixo', 'Médio', 'Alto', 'Topo'], duplicates='drop'
)

# (c) Média salarial por faixa
df.groupby('faixa_salario', observed=True)['salario'].mean().round(0)
```
