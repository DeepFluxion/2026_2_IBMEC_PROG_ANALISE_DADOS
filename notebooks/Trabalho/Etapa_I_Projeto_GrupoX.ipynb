## 7. Preparação e transformação dos dados

Para cada procedimento são apresentados: o código Python utilizado, um trecho do resultado obtido, uma breve explicação e a justificativa da transformação, quando aplicável.

### 7.1 Leitura da base de dados

A base é carregada diretamente da URL pública com a biblioteca **Pandas**. O arquivo é um CSV separado por vírgula, codificação UTF-8 e separador decimal padrão (ponto) — não foi necessário o uso dos parâmetros `sep`, `decimal` ou `encoding`.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

pd.set_option("display.max_columns", None)

URL = "https://raw.githubusercontent.com/DeepFluxion/2026_2_IBMEC_PROG_ANALISE_DADOS/refs/heads/main/datasets/melb_data.csv"

df = pd.read_csv(URL)

print(f"Quantidade de linhas:  {df.shape[0]}")
print(f"Quantidade de colunas: {df.shape[1]}")
df.head()
```

**Caracterização do dataframe (seção 6.1):**
- Nome do dataframe: `df` (`melb_data.csv`)
- Quantidade de linhas e colunas: ver saída da célula acima
- Origem: dados públicos de vendas de imóveis em Melbourne (ver seção 4)

### 7.2 Seleção e exclusão de colunas

| Coluna excluída | Justificativa |
|---|---|
| `Address` | Identificador de altíssima cardinalidade (praticamente único por linha); não agrega valor a análises agregadas e não é necessária para responder às questões de investigação. |
| `SellerG` | Nome da imobiliária/corretor responsável pela venda; não é objeto da análise proposta (foco em preço, localização e características do imóvel). |
| `Bedroom2` | Informação redundante em relação a `Rooms` (ambas registram o número de cômodos/quartos do imóvel, com alta correlação entre si); manter apenas `Rooms` evita duplicidade de informação. |

```python
colunas_excluidas = ["Address", "SellerG", "Bedroom2"]
df = df.drop(columns=colunas_excluidas)

print("Colunas removidas:", colunas_excluidas)
print(f"Quantidade de colunas após a exclusão: {df.shape[1]}")
df.columns.tolist()
```

### 7.3 Renomeação das colunas

As colunas foram renomeadas para `snake_case`, em minúsculas, sem espaços e com nomes mais descritivos (corrigindo também os erros de grafia originais `Lattitude` → `latitude`, `Longtitude` → `longitude`), seguindo as boas práticas apresentadas em aula.

```python
print("Colunas antes da renomeação:")
print(df.columns.tolist())

mapa_colunas = {
    "Suburb": "suburb",
    "Rooms": "rooms",
    "Type": "type",
    "Price": "price",
    "Method": "method",
    "Date": "date",
    "Distance": "distance",
    "Postcode": "postcode",
    "Bathroom": "bathroom",
    "Car": "car",
    "Landsize": "landsize",
    "BuildingArea": "building_area",
    "YearBuilt": "year_built",
    "CouncilArea": "council_area",
    "Lattitude": "latitude",
    "Longtitude": "longitude",
    "Regionname": "region_name",
    "Propertycount": "property_count",
}
df = df.rename(columns=mapa_colunas)

print("\nColunas depois da renomeação:")
print(df.columns.tolist())
```

**Boas práticas utilizadas:** nomes em `snake_case`, letras minúsculas, sem espaços ou acentos, correção de erros de grafia (`lattitude`/`longtitude`) e nomes autoexplicativos para cada variável.

### 7.4 Identificação e tratamento dos tipos de dados

Verificamos os tipos de dados atribuídos automaticamente pelo Pandas e corrigimos as inconsistências: `postcode` e `property_count` são lidos como `float64` mas representam valores inteiros/categóricos; `date` está como texto; colunas categóricas (`type`, `method`, `region_name`, `council_area`) são convertidas para o tipo `category`, mais eficiente em memória.

```python
print("Tipos de dados ANTES do tratamento:")
print(df.dtypes)
```

```python
# Postcode e property_count: valores inteiros armazenados como float -> Int64 (nullable)
df["postcode"] = df["postcode"].astype("Int64")
df["property_count"] = df["property_count"].astype("Int64")

# Colunas categóricas
for coluna in ["type", "method", "region_name", "council_area"]:
    df[coluna] = df[coluna].astype("category")

print("Tipos de dados DEPOIS do tratamento:")
print(df.dtypes)
```

### 7.5 Variáveis de data e hora

A coluna `date` contém a data da venda no formato texto `dia/mês/ano` (ex.: `3/12/2016`). Ela foi convertida para o tipo `datetime`, o que permite ordenar, filtrar e extrair componentes (ano, mês) da data de venda ao longo da análise.

```python
df["date"] = pd.to_datetime(df["date"], dayfirst=True)

print("Tipo de 'date' após a conversão:", df["date"].dtype)
print(f"Período coberto pela base: {df['date'].min().date()} a {df['date'].max().date()}")
df[["date"]].head()
```

### 7.6 Dados ausentes

Antes de tratar os valores ausentes, convertemos para `NaN` os valores de `building_area` iguais a **0**, pois uma área construída de 0 m² é fisicamente inconsistente e representa, na prática, um dado ausente disfarçado.

```python
df.loc[df["building_area"] == 0, "building_area"] = np.nan

ausentes_antes = df.isna().sum()
ausentes_antes_pct = (df.isna().mean() * 100).round(2)
resumo_ausentes_antes = pd.DataFrame({"qtd_ausentes": ausentes_antes, "pct_ausentes": ausentes_antes_pct})
print("Dados ausentes ANTES do tratamento:")
resumo_ausentes_antes[resumo_ausentes_antes["qtd_ausentes"] > 0]
```

**Técnica de tratamento escolhida para cada coluna, com justificativa:**

- **`car`** (≈0,5% ausentes, variável discreta): preenchido com a **moda** (valor mais frequente), pois a proporção de ausentes é muito pequena e a moda preserva a distribuição original da variável.
- **`council_area`** (≈10% ausentes, categórica): preenchido com a categoria **"Não informado"**, pois não há uma forma confiável de inferir o conselho administrativo a partir das demais colunas sem introduzir viés.
- **`building_area`** (≈48% ausentes, numérica contínua): proporção de ausentes muito alta para ser ignorada ou descartada (perderíamos quase metade da base). Optou-se pela **mediana agrupada por `type` e faixa de `rooms`**, que é mais robusta a outliers do que a média e respeita o fato de que o tamanho construído varia muito conforme o tipo de imóvel e o número de cômodos.
- **`year_built`** (≈39% ausentes, numérica discreta): preenchido com a **mediana por `suburb`** (e, quando o subúrbio não tiver nenhum valor conhecido, pela mediana geral da base), pois o ano de construção tende a ser semelhante entre imóveis do mesmo bairro.

```python
# car: moda
moda_car = df["car"].mode(dropna=True)[0]
df["car"] = df["car"].fillna(moda_car)

# council_area: categoria "Não informado"
if "Não informado" not in df["council_area"].cat.categories:
    df["council_area"] = df["council_area"].cat.add_categories(["Não informado"])
df["council_area"] = df["council_area"].fillna("Não informado")

# building_area: mediana por type + faixa de rooms
df["building_area"] = df.groupby(["type", "rooms"])["building_area"].transform(
    lambda s: s.fillna(s.median())
)
df["building_area"] = df["building_area"].fillna(df["building_area"].median())  # casos residuais

# year_built: mediana por suburb
df["year_built"] = df.groupby("suburb")["year_built"].transform(lambda s: s.fillna(s.median()))
df["year_built"] = df["year_built"].fillna(df["year_built"].median())  # casos residuais

ausentes_depois = df.isna().sum()
print("Dados ausentes DEPOIS do tratamento:")
print(ausentes_depois[ausentes_depois > 0] if ausentes_depois.sum() > 0 else "Nenhuma coluna com dados ausentes.")
```

### 7.7 Tratamento e padronização dos valores

Verificamos inconsistências de texto (maiúsculas/minúsculas, espaços extras) nas colunas categóricas e um erro de digitação identificado em `year_built`.

```python
# Padronização de espaços/caixa em colunas de texto
for coluna in ["suburb", "region_name"]:
    antes = df[coluna].astype(str)
    df[coluna] = df[coluna].astype(str).str.strip().str.title().astype("category" if coluna != "suburb" else "object")
    divergencias = (antes != df[coluna].astype(str)).sum()
    print(f"{coluna}: {divergencias} valores alterados pela padronização (espaços/caixa).")

print()

# Erro de digitação identificado em year_built (ex.: 1196 em vez de 1996)
print("Valores mínimo e máximo de year_built ANTES da correção:", df["year_built"].min(), "-", df["year_built"].max())
df.loc[df["year_built"] < 1800, "year_built"] = df.loc[df["year_built"] < 1800, "year_built"] + 800
print("Valores mínimo e máximo de year_built DEPOIS da correção:", df["year_built"].min(), "-", df["year_built"].max())
```

**Justificativa:** `suburb` e `region_name` foram padronizados com `strip()` (remoção de espaços extras nas bordas) e `title()` (capitalização consistente), evitando que o mesmo bairro apareça como categorias diferentes por causa de variações de escrita. Em `year_built`, identificamos valores muito abaixo do plausível (ex.: 1196), claramente erros de digitação de um dígito (faltando o "9" da casa das centenas); como o maior salto desse tipo observado corresponde a 800 anos, a correção soma 800 aos anos inválidos (< 1800), realinhando-os ao século correto.

### 7.8 Criação de novas variáveis

**7.8.1 — Nova variável numérica**
- **Nome da variável:** `price_per_m2`
- **Descrição:** preço de venda dividido pela área construída do imóvel (AUD por m²)
- **Finalidade no projeto:** permite comparar o valor dos imóveis de forma proporcional ao seu tamanho, em vez de apenas pelo preço absoluto.

```python
df["price_per_m2"] = (df["price"] / df["building_area"]).round(2)
df[["price", "building_area", "price_per_m2"]].head()
```

**7.8.2 — Nova variável categórica (criada com `.loc`)**
- **Nome da variável:** `faixa_distancia`
- **Descrição:** categoriza a distância do imóvel até o centro de Melbourne em quatro faixas
- **Finalidade no projeto:** facilita a comparação de preços e volume de vendas por proximidade ao centro (apoia a Questão de investigação 2).

```python
df["faixa_distancia"] = ""
df.loc[df["distance"] <= 5, "faixa_distancia"] = "Até 5 km"
df.loc[(df["distance"] > 5) & (df["distance"] <= 10), "faixa_distancia"] = "5 a 10 km"
df.loc[(df["distance"] > 10) & (df["distance"] <= 20), "faixa_distancia"] = "10 a 20 km"
df.loc[df["distance"] > 20, "faixa_distancia"] = "Acima de 20 km"
df["faixa_distancia"] = df["faixa_distancia"].astype("category")

df[["distance", "faixa_distancia"]].head()
```

### 7.9 Discretização de variáveis

Escolhida a variável numérica contínua **`price`** (preço de venda), discretizada em 4 faixas pela técnica **Equal Frequency** (frequência igual, via `pd.qcut`).

**Justificativa:** a distribuição de preços é assimétrica, com poucos imóveis de valor muito alto (outliers) "puxando" a cauda da distribuição. A técnica de Equal Width geraria faixas com quantidades de imóveis muito desbalanceadas; a Equal Frequency garante grupos de tamanho semelhante, mais adequados para comparações entre faixas de preço.

```python
df["faixa_preco"] = pd.qcut(
    df["price"],
    q=4,
    labels=["Q1 - Mais baixo", "Q2 - Médio-baixo", "Q3 - Médio-alto", "Q4 - Mais alto"],
)

df["faixa_preco"].value_counts().sort_index()
```

**Interpretação dos grupos formados:** cada uma das quatro faixas concentra aproximadamente 25% dos imóveis da base. A faixa "Q1 - Mais baixo" reúne os imóveis mais acessíveis e a "Q4 - Mais alto" os imóveis de maior valor — essa variável é usada nas tabelas de frequência e nas visualizações da seção 9 para comparar características entre faixas de preço.

### 7.10 Variáveis dummy

Variável categórica nominal escolhida: **`type`** (tipo do imóvel: `h` = casa, `u` = unidade/apartamento, `t` = sobrado), codificada em variáveis dummy (one-hot encoding).

```python
dummies_type = pd.get_dummies(df["type"], prefix="type")
df = pd.concat([df, dummies_type], axis=1)

print("Novas colunas criadas:", dummies_type.columns.tolist())
df[["type"] + dummies_type.columns.tolist()].head()
```

**Explicação das novas variáveis criadas:** `type_h`, `type_t` e `type_u` são colunas binárias (0/1) que indicam, respectivamente, se o imóvel é uma casa, um sobrado ou uma unidade/apartamento — formato necessário para uso da variável categórica em modelos estatísticos ou de machine learning.

- **Variáveis:** `price`, `type`
- **Tipo de gráfico:** boxplot
- **Justificativa:** o boxplot permite comparar a distribuição (mediana, dispersão e outliers) do preço entre as categorias de tipo de imóvel, respondendo diretamente à Questão de investigação 3.

```python
ordem_tipos = ["h", "t", "u"]
dados_boxplot = [df.loc[df["type"] == t, "price"] / 1_000_000 for t in ordem_tipos]

fig, ax = plt.subplots(figsize=(8, 5))
caixas = ax.boxplot(dados_boxplot, patch_artist=True)
ax.set_xticks([1, 2, 3])
ax.set_xticklabels(["Casa (h)", "Sobrado (t)", "Unidade (u)"])
for patch, cor in zip(caixas["boxes"], PALETA_CATEGORICA):
    patch.set_facecolor(cor)
    patch.set_alpha(0.75)
ax.set_title("Preço de venda por tipo de imóvel")
ax.set_xlabel("Tipo de imóvel")
ax.set_ylabel("Preço (milhões de AUD)")
plt.tight_layout()
plt.show()
```

**Interpretação:** casas (`h`) têm o maior preço mediano e a maior dispersão de valores, seguidas por sobrados (`t`); unidades/apartamentos (`u`) apresentam os menores preços e menor variabilidade — coerente com o maior tamanho médio das casas.

**Gráfico 3 — Preço médio por região**
- **Variáveis:** `price`, `region_name`
- **Tipo de gráfico:** barras
- **Justificativa:** gráfico de barras é adequado para comparar uma medida (preço médio) entre categorias (regiões), respondendo à Questão de investigação 1.

```python
preco_medio_regiao = (df.groupby("region_name", observed=True)["price"].mean() / 1_000_000).sort_values(ascending=False)

fig, ax = plt.subplots(figsize=(9, 5))
ax.bar(preco_medio_regiao.index, preco_medio_regiao.values, color=COR_PRINCIPAL)
ax.set_title("Preço médio de venda por região")
ax.set_xlabel("Região")
ax.set_ylabel("Preço médio (milhões de AUD)")
plt.xticks(rotation=40, ha="right")
plt.tight_layout()
plt.show()
```

**Interpretação:** a região Southern Metropolitan apresenta o maior preço médio, seguida pela Eastern Metropolitan; as regiões "Victoria" (mais afastadas do centro) têm os menores preços médios — um primeiro indício da relação entre localização e preço.

**Gráfico 4 — Relação entre distância ao centro e preço**
- **Variáveis:** `distance`, `price`
- **Tipo de gráfico:** dispersão (scatter)
- **Justificativa:** o gráfico de dispersão é o mais adequado para investigar a relação entre duas variáveis numéricas contínuas, respondendo diretamente à Questão de investigação 2.

```python
amostra = df.sample(n=2000, random_state=42)

fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(amostra["distance"], amostra["price"] / 1_000_000, color=COR_PRINCIPAL, alpha=0.4, s=18)
ax.set_title("Distância ao centro x Preço de venda")
ax.set_xlabel("Distância ao centro (km)")
ax.set_ylabel("Preço (milhões de AUD)")
plt.tight_layout()
plt.show()
```

**Interpretação:** observa-se uma leve tendência de queda do preço conforme a distância ao centro aumenta, mas a relação não é fortemente linear — há imóveis de alto valor mesmo em distâncias maiores, indicando que outros fatores (região, tipo, área) também pesam na formação do preço.

**Gráfico 5 — Quantidade de imóveis por faixa de distância**
- **Variáveis:** `faixa_distancia`
- **Tipo de gráfico:** barras
- **Justificativa:** complementa o Gráfico 4, mostrando o volume de vendas em cada faixa de distância criada na seção 7.8.

```python
contagem_faixa = df["faixa_distancia"].value_counts().reindex(
    ["Até 5 km", "5 a 10 km", "10 a 20 km", "Acima de 20 km"]
)

fig, ax = plt.subplots(figsize=(8, 5))
ax.bar(contagem_faixa.index, contagem_faixa.values, color=PALETA_CATEGORICA[1])
ax.set_title("Quantidade de imóveis vendidos por faixa de distância ao centro")
ax.set_xlabel("Faixa de distância")
ax.set_ylabel("Quantidade de imóveis")
plt.tight_layout()
plt.show()
```

**Interpretação:** a maior parte das vendas concentra-se nas faixas de 5 a 20 km do centro, região que reúne grande parte dos bairros residenciais consolidados de Melbourne.
