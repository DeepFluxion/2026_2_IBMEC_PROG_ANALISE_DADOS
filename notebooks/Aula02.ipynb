## Estruturas de Controle

### Condicionais (o que são)

Servem para **tomada de decisão** no código.
- Executam blocos **se** uma condição booleana for verdadeira.
- Sequenciam regras de negócio (ex.: preço, imposto, classificação).

````python
receita = 120000
custo = 95000

if receita > custo:
    print("Resultado: Lucro")
````

# if/else e elif (explicação)

- `if`: testa a 1ª condição.
- `elif`: condições adicionais, em ordem.
- `else`: “caso contrário” (pega o restante).
Apenas **um** bloco executa na cadeia.

````python
# if/elif/else (exemplo: variação do PIB)
variacao_pib = 1.7  # em %

if variacao_pib >= 2.5:
    classificacao = "Aquecido"
elif 0 <= variacao_pib < 2.5:
    classificacao = "Estável"
else:
    classificacao = "Recessivo"

print("Classificação econômica:", classificacao)
````

# Operadores lógicos e operador ternário

- Comparação: `==`, `!=`, `>`, `>=`, `<`, `<=`
- Lógicos: `and`, `or`, `not`
- **Ternário**: `expr_if_true if condição else expr_if_false`
Úteis para combinar regras (preço, demanda, estoque etc.).


````python
preco = 150
demanda = 40
cliente_fidelidade = True

# Combinação lógica
if (preco > 100) and (demanda < 50):
    print("Sinal de queda de vendas: preço alto e demanda baixa.")

# Ternário para desconto
desconto = 0.10 if cliente_fidelidade else 0.0
preco_final = preco * (1 - desconto)
print(f'Preço Inicial R${preco} | Preço final: R${preco_final}')
````

# match / case (pattern matching)

Permite padronizar múltiplos casos de forma legível.
- Substitui cadeias longas de `elif`.
- Suporta **guards** (condições) com `if`.

````python
inflacao = 4.3  # em %

match inflacao:
    case x if x < 3:
        faixa = "Baixa"
    case x if 3 <= x <= 6.5:
        faixa = "Meta / Controlada"
    case _:
        faixa = "Alta"

faixa
````

# Laços: visão geral

- `for`: percorre coleções ou faixas (quando sabemos “quantos”).
- `while`: repete **enquanto** a condição for verdadeira (útil quando não sabemos previamente o número de iterações).
Atenção com **condições de parada** para evitar loops infinitos.


````python
# for
vendas = [100, 120, 90, 140, 180]

# Percorrendo com índice e valor
for i, valor in enumerate(vendas, start=1):
    print(f"Mês {i}: {valor}")

# Compreensão de listas
preco_sem_imposto = [10, 20, 30]
preco_com_imposto = [p * 1.07 for p in preco_sem_imposto]  # 7% de imposto
preco_com_imposto
````

# while

Repete um bloco **enquanto** a condição for verdadeira.
Use quando:
- Depende de um estado que muda “durante” o loop (ex.: estoque).
- Número de iterações é desconhecido a priori.
Sempre **atualize** as variáveis para garantir parada.


````python
estoque = 7
vendas_por_ciclo = [2, 1, 3, 2]
i = 0

while estoque > 0 and i < len(vendas_por_ciclo):
    estoque -= vendas_por_ciclo[i]
    i += 1
    print("Estoque restante:", max(estoque, 0))  # evita negativo na impressão

# Exemplo com condição de parada explícita
contador = 0
limite = 5
while True:            # laço “até parar”
    contador += 1
    if contador >= limite:
        break          # condição de parada
contador
````

# break, continue, pass (explicação)

- `break`: interrompe o laço imediatamente.
- `continue`: pula para a **próxima** iteração.
- `pass`: “não faz nada” (placeholder).

````python
# Slide 7.14 — break, continue, pass (exemplos)
transacoes = [100, -20, 300, 0, 50, -10]
saldo = 0

for t in transacoes:
    if t == 0:
        pass  # placeholder: nada a fazer para zero
        continue  # pula direto para a próxima
    saldo += t
    if saldo < 0:
        print("Alerta: saldo negativo! Encerrando análise.")
        break

saldo
````
